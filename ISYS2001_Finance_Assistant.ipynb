Smart Personal Budget Assisstant

Step 1 Understand the Problem
Target Users:
University student or young teenagers who want to mange their monthly income and expenses.

Business problem:
Many university students have limited income and struggle to manage their monthly expenses, including rental, food expenses, tranportation, and entertainment. Without a clear overview and managemnet of their spending, students may overspend and difficult for saving.

Proposed Solution:
The Smart Personal Budget Assisstant is a application designed to help students mange their personal finances.

The application allows users to enter their monthly income and upload a csv file containing their financial transactions. It calculates total expenses, remaining balances,and spending percentages, to identities the largest spending categories.

The application also uses Google's Gemini AI to answer finance-related question based on the user's actual financiual data.

Project Objectives:
1. Calculate montly expenses and remianing balance.
2. Analyse financial trasaction from a CSV file.
3. Identity the largest spending categories.
4. Provide personalised financial explanation usingGemini AI.

Expected benefit:
The application helps users understand their spending patterns, manage their montly budgets, and make more informed financial decisions.




Step 2 Identity Inputs and Outputs

Input 
The Smart Personal Budget assisstant requires the following information from users.

1. Montly income: User's total monthly income
2. CSV File: Contains the user's financial transaction.
3. Date: The date of each transactions
4. Description: The details of each transactions
5. Category: The expense category, such as rental, food, entertainment.
6. Amount: The transaction amount.
7. User questions: The question ask by user

Output
1. Total expenses: The total amount spend during the period
2. Remaining Balance: Monthly income minus total expenses
3. Spending percentage: The percentage of monthly income spend
4. Category breakdown: Total spend in each category.
5. Largest expenses category: The category with highest spending
6. Ai Financial Response: A personalised financial explanation generated by Gemini AI.

Data Validation
The application will caheck the monthly income is valid and non-negative, the csv file contains teh required columns, and the transaction amount is numeric, The invalid inputs will show a error messages.

Step 3: Worked Example by Hand

Example Scenario

A university student earns AUD 3,000 per month and has the following expenses:

* Rent: AUD 1,000
* Food: AUD 500
* Transport: AUD 200
* Entertainment: AUD 300

Calculation 1: Total Expenses

Total Expenses = Rent + Food + Transport + Entertainment
= 1,000 + 500 + 200 + 300
= AUD 2,000

Calculation 2: Remaining Balance

Remaining Balance = Monthly Income − Total Expenses
= 3,000 − 2,000
= AUD 1,000

Calculation 3: Spending Percentage

Spending Percentage = (Total Expenses / Monthly Income) × 100
= (2,000 / 3,000) × 100
= 66.67%

Calculation 4: Largest Expense Category

The largest expense category is Rent, with AUD 1,000.
Rent Percentage = (Rent / Total Expenses) × 100
= (1,000 / 2,000) × 100
= 50%

Expected Output

1. Monthly Income: AUD 3,000
2. Total Expenses: AUD 2,000
3. Remaining Balance: AUD 1,000
4. Spending Percentage: 66.67%
5. Largest Expense Category: Rent

Interpretation

The student spends 66.67% of their monthly income and has AUD 1,000 remaining. Rent is the largest expense category, accounting for 50% of total expenses. The application can help the student understand their spending habits and identify opportunities to save money.

```python
##Step 5 Python implementation

# Initial budget calculation

# Function 1: Calculate total expenses
def calculate_total_expenses(expenses):
    total = sum(expenses.values())
    return total


# Function 2: Calculate remaining balance
def calculate_remaining_balance(income, total_expenses):
    remaining = income - total_expenses
    return remaining


# Function 3: Calculate spending percentage
def calculate_spending_percentage(total_expenses, income):
    if income == 0:
        return 0

    percentage = (total_expenses / income) * 100
    return percentage


# Function 4: Find the largest expense category
def find_largest_category(expenses):
    if not expenses:
        return "None"

    largest = max(expenses, key=expenses.get)
    return largest


# Example financial data
monthly_income = 3000

expenses = {
    "Rent": 1000,
    "Food": 500,
    "Transport": 200,
    "Entertainment": 300
}


# Perform calculations
total_expenses = calculate_total_expenses(expenses)

remaining_balance = calculate_remaining_balance(
    monthly_income, total_expenses
)

spending_percentage = calculate_spending_percentage(
    total_expenses, monthly_income
)

largest_category = find_largest_category(expenses)


# Display results
print("SMART PERSONAL BUDGET ASSISTANT")
print("------------------------------")

print(f"Monthly Income: AUD {monthly_income:.2f}")
print(f"Total Expenses: AUD {total_expenses:.2f}")
print(f"Remaining Balance: AUD {remaining_balance:.2f}")
print(f"Spending Percentage: {spending_percentage:.2f}%")
print(f"Largest Expense Category: {largest_category}")
```

```python
# Build the Gradio User Interface (GUI) for Smart Personal Budget Assistant.

# Installing Gradio 
%pip install gradio

# Gradio User Interface (GUI) Implementation
import gradio as gr
import pandas as pd

def budget_assistant(income, csv_file, question):

    # Check user inputs
    if csv_file is None:
        return "Please upload your expense CSV file."

    if income is None or income < 0:
        return "Please enter a valid monthly income."

    if not question or not question.strip():
        return "Please enter a financial question."

    try:
        # Read uploaded CSV
        df = pd.read_csv(csv_file)

        # Check required columns
        required_columns = ["Date", "Description", "Category", "Amount"]

        for column in required_columns:
            if column not in df.columns:
                return f"Missing required column: {column}"

        # Validate amounts
        df["Amount"] = pd.to_numeric(df["Amount"], errors="raise")

        if df["Amount"].isna().any():
            return "Expense amounts cannot be empty."

        if (df["Amount"] < 0).any():
            return "Expense amounts cannot be negative."

        # Calculate financial information
        total_expenses = df["Amount"].sum()
        remaining_balance = income - total_expenses

        category_totals = df.groupby("Category")["Amount"].sum()

        if income > 0:
            spending_percentage = (total_expenses / income) * 100
            percentage_text = f"{spending_percentage:.2f}%"
        else:
            percentage_text = "N/A (income is zero)"

        if category_totals.empty:
            largest_category = "None"
        else:
            largest_category = category_totals.idxmax()

        # Prepare budget information for Gemini
        financial_summary = f"""
Monthly Income: AUD {income:.2f}
Total Expenses: AUD {total_expenses:.2f}
Remaining Balance: AUD {remaining_balance:.2f}
Spending Percentage: {percentage_text}
Largest Expense Category: {largest_category}

Expenses by Category:
{category_totals.to_string()}
"""

        # Ask Gemini using the existing finance chatbot
        prompt = f"""
Here is the student's financial information:

{financial_summary}

Student's question:
{question}

Answer using the student's actual financial data.
Do not invent income or expense figures.
"""

        advice = finance_chatbot(prompt)

        # Return the result
        result = f"""
MONTHLY BUDGET SUMMARY

{financial_summary}

GEMINI FINANCIAL ADVICE

{advice}
"""

        return result

    except Exception as error:
        return f"Error: {error}"


# Create Gradio GUI
with gr.Blocks(title="Smart Personal Budget Assistant") as demo:

    gr.Markdown("# Smart Personal Budget Assistant")
    gr.Markdown(
        "Upload your expenses, enter your monthly income, "
        "and ask Gemini for personalised budgeting advice."
    )

    income_input = gr.Number(
        label="Monthly Income (AUD)",
        value=3000,
        minimum=0
    )

    csv_input = gr.File(
        label="Upload Expense CSV",
        file_types=[".csv"],
        type="filepath"
    )

    question_input = gr.Textbox(
        label="Ask a Financial Question",
        placeholder="How can I reduce my spending?",
        lines=2
    )

    submit_button = gr.Button("Analyse My Budget")

    output_box = gr.Textbox(
        label="Budget Analysis and Gemini Advice",
        lines=18
    )

    submit_button.click(
        fn=budget_assistant,
        inputs=[income_input, csv_input, question_input],
        outputs=output_box
    )

demo.launch(share=False)

# Gradio Application Testing - test with each scenario if providing error messsage 
# Example: 1. test with normal version positive income, upload csv, and ask a question /2. test with negative income, upload csv and ask a question. The gradio must show error and ask for correct information if the input is invalid.
```
